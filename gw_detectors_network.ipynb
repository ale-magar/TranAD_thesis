```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import numpy as np

# Detector positions
detectors = {
    "H1": (46.45, -119.41, "blue"),
    "L1": (30.56, -90.77, "red"),
    "Virgo": (43.63, 10.50, "purple"),
    "KAGRA": (36.41, 137.31, "green"),
}

# Figure
fig = plt.figure(figsize=(12, 6))
ax = plt.axes(projection=ccrs.PlateCarree())
ax.set_global()

# Coastlines and borders
ax.coastlines(linewidth=0.8)
ax.add_feature(cfeature.BORDERS, linewidth=0.4)

# ---- TICKS EVERY 20 DEGREES ----
xticks = np.arange(-180, 181, 20)
yticks = np.arange(-80, 81, 20)

ax.set_xticks(xticks, crs=ccrs.PlateCarree())
ax.set_yticks(yticks, crs=ccrs.PlateCarree())

ax.tick_params(labelsize=10)

# Gridlines (dotted like thesis figure)
ax.grid(True, linestyle=":", linewidth=0.6)

# Axis labels (MATCHING YOUR PDF)
ax.set_xlabel(r"phi", fontsize=12)
ax.set_ylabel(r"theta", fontsize=12)

# Plot detectors
for name, (lat, lon, color) in detectors.items():
    ax.plot(lon, lat, "o", color=color, markersize=8,
            transform=ccrs.PlateCarree())
    ax.text(lon + 3, lat + 2, name, fontsize=11, color=color,
            transform=ccrs.PlateCarree())

plt.tight_layout()
plt.savefig("gw_detectors_network.png")
plt.show()
```
